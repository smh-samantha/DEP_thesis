### Framework Parameter Logic

The framework defines all aircraft, control, propulsion, and performance quantities as parameters.

Each parameter has:

1. a baseline value
2. an optional range of values
3. a role in the current study

Possible roles:

- fixed input
- swept variable
- output
- feasibility limit
- optimisation objective

A parameter can change role depending on the design question.

Example:

- motor thrust limit can be fixed if the motor is already chosen
- motor thrust limit can be swept if the framework is sizing the motor
- motor thrust utilisation can be an output
- maximum allowed utilisation can be a feasibility limit

### 3. Aircraft Dynamics Model

The aircraft dynamics model converts the framework parameters into a mathematical model of how the aircraft responds to disturbances and control inputs.

This step produces the model used by the controller, stability sweep, control effort calculation, and robustness checks.

### 3.1 Required Parameters

The model needs parameters that describe aircraft mass, flight condition, aerodynamic stability, and control effectiveness.

Required variables:

- `m` = aircraft mass
- `Iyy` = pitch moment of inertia
- `V` = flight speed
- `rho` = air density
- `S` = wing reference area
- `c_bar` = mean aerodynamic chord
- `Cm_alpha` = pitching moment change due to angle of attack
- `Cm_q` = pitching moment change due to pitch rate
- `Cm_delta` = pitching moment change due to control input
- `CL_alpha` = lift change due to angle of attack
- `CL_delta` = lift change due to control input

### 3.2 Model Form

The first aircraft dynamics model will use a reduced longitudinal state-space model.

State vector: `x = [alpha, q, theta]`

Control input: `u = pitch control input`

State-space form: `x_dot = A x + B u`

### 3.3 Outputs

This step outputs:

- `A` matrix
- `B` matrix
- open-loop eigenvalues
- dominant pole
- open-loop stability classification
- baseline disturbance response

### 3.4 What This Step Answers

This step answers:

- is the aircraft naturally stable?
- how quickly does a disturbance grow or decay?
- how does reduced passive stability affect the open-loop aircraft?
- what model will the controller need to stabilise?

### 3.5 Validation / Checks

Checks for this step:

- baseline aircraft behaves like a reasonable reference aircraft
- stable cases have eigenvalues with negative real parts
- reducing passive stability moves eigenvalues toward instability
- results are compared against Thesis A and at least one published/reference aircraft model












### Reference Model Direction

Use the Vechtel/Buch DEP yaw-control paper as the main framework benchmark.

The framework will focus on lateral-directional stability and yaw control.

Main comparison cases:

1. 2-propeller configuration
2. 6-propeller configuration
3. 12-propeller configuration
4. reduced vertical tail plane cases

The framework should test:

1. how VTP reduction changes directional stability
2. how DEP layout changes available yaw control authority
3. how much differential thrust is required
4. whether actuator limits are exceeded
5. whether robustness and handling degrade
6. where the feasible / infeasible boundary occurs

The goal is not to fully recreate their 6DOF simulator.

The goal is to build a lower-order framework that can reproduce the same broad trade-off logic:

VTP reduction
→ reduced passive yaw stability
→ increased need for rudder and differential thrust
→ changed control effort, energy use, robustness, and feasibility
